### 문제 2

현재 주어진 데이터 셋으로 어떤 비지니스 지표를 만들 수 있는지 제시하고, 해당 지표들로 AARRR 프레임 워크를 구성할 수 있는지 판단

##### Acquisition(획득)

- 신규 고객 유입수

```python
# 주문 데이터에서 고객 고유 ID와 주문 구매 시점 추출 및 정렬, 중복된 custoer_id 첫 주문 외 제거
customer_order_first = orders_df[['customer_id', 'order_purchase_timestamp']].sort_values(by='order_purchase_timestamp')
```

```python
customer_order_first = customer_order_first.drop_duplicates(subset='customer_id', keep='first')
```

```python
customers_df2 = pd.merge(customers_df, customer_order_first, on='customer_id', how='left')
```

```python
customers_df2['first_purchase_time'] = pd.to_datetime(customers_df2['order_purchase_timestamp'])
```

```python
# 월별 신규 고객 추이 시각화
customers_df2['first_purchase_month'] = customers_df2['first_purchase_time'].dt.to_period('M')
new_customers_M = customers_df2['first_purchase_month'].value_counts().sort_index()
plt.figure(figsize=(12, 6))
sns.lineplot(x=new_customers_M.index.astype(str), y=new_customers_M.values)
plt.title('신규 고객 추이(월)')
plt.xlabel('월')
plt.ylabel('신규 고객 수')
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# 일별 신규 고객 추이 시각화
customers_df2['first_purchase_month'] = customers_df2['first_purchase_time'].dt.to_period('D')
new_customers_D = customers_df2['first_purchase_month'].value_counts().sort_index()
plt.figure(figsize=(12, 6))
sns.lineplot(x=new_customers_D.index.astype(str), y=new_customers_D.values)
plt.title('신규 고객 추이(일)')
plt.xlabel('일')
plt.ylabel('신규 고객 수')
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()
```

- 위 그래프를 이용해 이상치로 보이는 값들의 시기와 해당 시기의 이벤트를 대조하여 해당 시기에 이벤트가 어느정도의 성과가 있었는지를 확인하여 마케팅 전략을 기획하는데 이용한다면 activation 단계로 자연스레 이어지고 수익성 또한 올릴 수 있을 것이라 예상된다. 특히 17년 12월 달의 데이터를 자세히 살펴볼 필요성이 큼.

##### Retention(유지)

```python
orders_reviews_df = pd.merge(order_reviews_df, orders_df[['order_id', 'order_purchase_timestamp']], on='order_id', how='left')
```

```python
orders_reviews_df['order_purchase_timestamp'] = pd.to_datetime(orders_reviews_df['order_purchase_timestamp'])
```

```python
orders_reviews_df['order_purchase_month'] = orders_reviews_df['order_purchase_timestamp'].dt.to_period('M')
monthly_avg_review_score_purchase = orders_reviews_df.groupby('order_purchase_month')['review_score'].mean().sort_index()
```

```python
# 시각화 (주문 구매 시점 기준)
plt.figure(figsize=(12, 6))
sns.lineplot(x=monthly_avg_review_score_purchase.index.astype(str), y=monthly_avg_review_score_purchase.values)
plt.title('월별 평균 리뷰 점수 추이 (주문 구매 시점 기준)')
plt.xlabel('월')
plt.ylabel('평균 리뷰 점수')
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()
```

저조한 성적의 월의 배송지연, 상품파손 등과의 원인을 분석하여 비교를 추가적으로 진행하여 사용자의 만족도를 보다 충족시킬 수 있는 방향성을 제시
데이터 특성상 고객 유지에 대한 데이터를 확인하기 어려움. 따라서 판매량을 유지시키는 방향으로 진행하여 리뷰를 통해 사용자의 만족도를 파악해 서비스 제공에 퀄리티를 올리는 것을 제시.

##### Referral(추천)

```python
# 리뷰의 점수가 높은 사람들을 추려 파악하고자함.
reviews_df2 = pd.merge(order_reviews_df, orders_df, on='order_id', how='left')
reviews_df3 = pd.merge(reviews_df2, customers_df, on='customer_id', how='left')
```

```python
reviews_df3.info()
```

```python
#리뷰 4점 이상
satisfied_customers_reviews = reviews_df3[reviews_df3['review_score'] >= 4]
satisfied_customer_ids = satisfied_customers_reviews['customer_unique_id'].unique()
num_satisfied_customers = len(satisfied_customer_ids)
print(f"4점 이상 리뷰를 남긴 고객 수: {num_satisfied_customers}")
```

```python
# 전체 고객 수
total_unique_customers = customers_df['customer_unique_id'].nunique()
print(f"전체 고객 수: {total_unique_customers}")
```

```python
# 만족한 고객 비율
satisfied_customer_ratio = (num_satisfied_customers / total_unique_customers) * 100
print(f"전체 고객 중 4점 이상 리뷰를 남긴 고객 비율: {satisfied_customer_ratio:.2f}%")
```

데이터상에 정확한 추천 관련 정보는 없어 정확한 분석은 어려우나 만족도 4점 이상의 비율이 76.83%로 만족도 4점이상의 사람이 추천한다는 가설을 세워보면 약76%의 고객이 신규고객에게 서비스를 추천한다 가정할 수 있다.
