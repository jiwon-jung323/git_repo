# 1. 🚲프로젝트 소개🚲

- 서울시 공공자전거인 ‘따릉이’는 2015년 10월부터 정식 운영을 시작한 서울시 공공자전거 대여 서비스입니다.\
서울교통빅테이터플랫폼인 T Data의 자료를 살펴보면, 자치구별로 이용량의 큰 차이가 있음을 파악할 수 있습니다.\
본 프로젝트는 자치구별 따릉이의 이용량 차이를 살펴보고, 자전거 이용과 관련된 자치구별 인프라(대여소 개수, 자전거도로면적 등), 시설(지하철역, 사업체)등을 비교합니다.\
그로 인해 자치구별 이용량 차이의 발생 원인을 파악하는 것을 목표로 하였습니다.

# 2. 배경✒️

따릉이_이용_통계.PNG

- 서울시에서 이용가능한 자전거 대여서비스인 '따릉이'의 통계에 따르면 서울시 자치구별 공공자전거 이용량이 강서구가 가장 많고, 강북구가 가장 적은 것을 알 수 있습니다. 위 자료를 토대로 자치구별 서울시 공공자전거 이용량의 차이가 큰 이유를 알아보고자 서울시의 자전거 이용과 관련된 데이터들을 활용하여 프로젝트를 진행하였습니다.
